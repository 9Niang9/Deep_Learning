## 3. Les tenseurs : les briques de base

**Qu'est-ce qu'un tenseur ?**

Un tenseur est un tableau multidimensionnel, similaire aux tableaux NumPy mais doté de capacités supplémentaires :
- Peut s'exécuter sur des GPU pour des calculs plus rapides
- Prend en charge la différentiation automatique (essentielle pour entraîner des réseaux de neurones)

**Dimensions des tenseurs :**
- Tenseur 0D (scalaire) : `5`
- Tenseur 1D (vecteur) : `[1, 2, 3]`
- Tenseur 2D (matrice) : `[[1, 2], [3, 4]]`
- Tenseur 3D+ : tableaux de dimension supérieure (p. ex. images, vidéos)

## 6. Vectorisation : écrire du code efficace

**Qu'est-ce que la vectorisation ?**

La vectorisation consiste à effectuer des opérations sur des tableaux entiers en une seule fois, plutôt que d'utiliser des boucles. Elle est :
- **Plus rapide** : elle exploite des implémentations bas niveau optimisées
- **Plus lisible** : moins de code, plus facile à lire
- **Plus Pythonique** : elle suit les bonnes pratiques

Voyons la différence :


````python
import time

# Approche non vectorisée (LENTE)
x = torch.tensor([1, 2, 3, 4, 5])

start = time.time()
for i in range(len(x)):
    x[i] = x[i] + 5

time_loop = time.time() - start

print("Résultat non vectorisé :", x)

print(f"Temps : {time_loop:.6f} secondes")
````

````python
# Approche vectorisée (RAPIDE)
x = torch.tensor([1, 2, 3, 4, 5])

start = time.time()

x = x + 5

time_vec = time.time() - start

print("Résultat vectorisé :", x)

print(f"Temps : {time_vec:.6f} secondes")

print(f"\nAccélération : {time_loop/time_vec:.1f}x plus rapide !")
````

### 6.1 Opérations vectorielles courantes

Implémentons des opérations fondamentales qui constituent les briques de base des réseaux de neurones.

#### Produit scalaire

Le produit scalaire de deux vecteurs $\mathbf{x}$ et $\mathbf{y}$ est :
$$\mathbf{x} \cdot \mathbf{y} = \sum_{i=1}^{n} x_i y_i = x_1y_1 + x_2y_2 + \cdots + x_ny_n$$

**Cas d'utilisation en apprentissage profond :**
- Calculer la similarité entre des vecteurs
- Multiplication matricielle (combinaison de produits scalaires)
- Mécanismes d'attention


````python
def dot_product(x, y):
    """
    Calcule le produit scalaire de deux tenseurs 1D.

    Arguments :
        x: Tenseur 1D
        y: Tenseur 1D (même taille que x)

    Renvoie :
        Tenseur scalaire représentant le produit scalaire
    """
    return torch.dot(x, y)

````

````python
# Exemple
x = torch.tensor([1.0, 2.0, 3.0])
y = torch.tensor([4.0, 5.0, 6.0])
result = dot_product(x, y)
print(f"x · y = {result}")
````

#### Produit extérieur

Le produit extérieur des vecteurs $\mathbf{x}$ (taille $m$) et $\mathbf{y}$ (taille $n$) produit une matrice :
$$\mathbf{x} \otimes \mathbf{y} = \mathbf{x}\mathbf{y}^T = \begin{bmatrix} x_1y_1 & x_1y_2 & \cdots \\ x_2y_1 & x_2y_2 & \cdots \\ \vdots & \vdots & \ddots \end{bmatrix}$$

**Utilisation en apprentissage profond :** calcul des poids d'attention, couches bilinéaires


````python
def PYTORCH_outer(x, y):
    """
    Calcule le produit extérieur de deux vecteurs.

    Paramètres :
        x (torch.Tensor): Tenseur 1D of shape (m,)
        y (torch.Tensor): Tenseur 1D of shape (n,)

    Renvoie :
        torch.Tenseur : Tenseur 2D de forme (m, n)
    """
    # À FAIRE : implémenter avec torch.outer() ou un remodelage suivi d’une multiplication
    # Indice : torch.outer(x, y) ou x.unsqueeze(1) * y.unsqueeze(0)
    return torch.outer(x, y)  # Remplacez cette ligne par votre implémentation
````

#### Produit élément par élément (Hadamard)

Multiplication élément par élément : $\mathbf{z} = \mathbf{x} \odot \mathbf{y}$ où $z_i = x_i \times y_i$

**Utilisation en apprentissage profond :** mécanismes de porte (LSTM, GRU), masquage d'attention


````python
def PYTORCH_multiply(x, y):
    """
    Multiplie les arguments élément par élément.

    Paramètres :
        x (torch.Tensor): Tenseur d’entrée
        y (torch.Tensor): Tenseur d’entrée (same shape as x)

    Renvoie :
        torch.Tenseur : Produit élément par élément
    """
    # À FAIRE : implémenter avec l’opérateur * ou torch.mul()
    return torch.mul(x,y)
````

#### Fonction d'activation ReLU

ReLU (unité linéaire rectifiée) est l'une des fonctions d'activation les plus populaires :
$$\text{ReLU}(x) = \max(0, x) = \begin{cases} x & \text{si } x > 0 \\ 0 & \text{si } x \leq 0 \end{cases}$$

**Pourquoi ReLU ?**
- Aide à résoudre le problème de disparition du gradient
- Efficace en calcul
- Introduit de la non-linéarité

**Illustration :**
- Entrée : `[-2, -1, 0, 1, 2]`
- Sortie : `[0, 0, 0, 1, 2]`


````python
def PYTORCH_ReLU(x):
    """
    Applique la fonction unité linéaire rectifiée élément par élément.

    Paramètres :
        x (torch.Tensor): Tenseur d’entrée

    Renvoie :
        torch.Tenseur : ReLU appliquée élément par élément
    """
    # À FAIRE : implémenter avec torch.relu() ou torch.maximum()
    # Indice : torch.relu(x) ou torch.maximum(x, torch.tensor(0.0))
    return torch.maximum(x, torch.tensor(0.0))
````

## 8. Différentiation automatique (Autograd)

**Pourquoi Autograd est essentiel :**

L'entraînement des réseaux de neurones exige le calcul de gradients (dérivées) pour la rétropropagation. Le système autograd de PyTorch s'en charge automatiquement !

**Concepts clés :**
- `requires_grad=True` : indique à PyTorch de suivre les opérations
- `.backward()` : calcule automatiquement les gradients
- `.grad` : stocke le gradient calculé

### 8.1 Exemple simple

Calculons $\frac{dy}{dx}$ où $y = x^2$


````python
# Calcul simple de gradient
x = torch.tensor(2.0, requires_grad=True)  # Suivre ce tenseur
y = x**2  # y = 4.0
````

````python
print(f"y = {y}")
print(f"dy/dx analytiquement = 2x = {2 * 2}")
````

````python




# Calculer le gradient
y.backward()  # Compute dy/dx
print(f"dy/dx (calculé) = {x.grad}")
````

**Que s'est-il passé ?**
1. Nous avons créé `x` avec `requires_grad=True`
2. PyTorch a suivi toutes les opérations sur `x` (sa mise au carré)
3. `.backward()` a calculé automatiquement le gradient
4. Le résultat est stocké dans `x.grad`


### 8.2 Exemple plus complexe

Calculons les gradients de : $y = a^2 + c \cdot b$

Alors :
- $\frac{\partial y}{\partial a} = 2a$
- $\frac{\partial y}{\partial b} = c$
- $\frac{\partial y}{\partial c} = b$


````python
a = torch.tensor(2.0, requires_grad=True)
b = torch.tensor(4.0, requires_grad=True)
c = torch.tensor(3.0, requires_grad=True)

y = a**2 + c * b
print(f"y = {y}")

y.backward()

print(f"\nGradients :")
print(f"∂y/∂a = {a.grad} (attendu : 2*2 = 4)")
print(f"∂y/∂b = {b.grad} (attendu : c = 3)")
print(f"∂y/∂c = {c.grad} (attendu : b = 4)")
````

### 8.3 Graphe de calcul

PyTorch construit un **graphe de calcul** qui suit toutes les opérations :

```
     a           b        c
     |           |        |
   square       mul ------+
     |           |
     +----add----+
          |
          y
```

Pendant `.backward()`, PyTorch parcourt ce graphe en sens inverse pour calculer les gradients.


````python
# Inspecter le graphe de calcul
x = torch.tensor(2.0, requires_grad=True)
y = x**2 + 3*x + 1

print("y.grad_fn:", y.grad_fn)  # Affiche la dernière opération
print("\nStructure du graphe de calcul :")
print("y.grad_fn.next_functions:", y.grad_fn.next_functions)
````

### 8.4 ⚠️ Important : les gradients s'accumulent !

Par défaut, appeler `.backward()` plusieurs fois **additionne** les gradients au lieu de les remplacer.


````python
# Accumulation de gradients (généralement indésirable !)
x = torch.tensor(2.0, requires_grad=True)

for step in range(3):
    y = x**2
    y.backward()  # Les gradients s’accumulent !
    print(f"Étape {step+1}: x.grad = {x.grad}")
````

**Solution :** effacez les gradients avant chaque passe arrière :


````python
# Gestion correcte des gradients
x = torch.tensor(2.0, requires_grad=True)

for step in range(3):
    if x.grad is not None:
        x.grad.zero_()  # Effacer les gradients précédents

    y = x**2
    y.backward()
    print(f"Étape {step+1}: x.grad = {x.grad}")
````

**En pratique :**
```python
# Training loop pattern
optimizer.zero_grad()  # Clear gradients
output = model(input)  # Forward pass
loss = criterion(output, target)
loss.backward()        # Compute gradients
optimizer.step()       # Update weights
```

### 9.3 Définir le réseau de neurones

Nous allons créer un réseau de neurones simple à propagation avant avec :
- Couche d'entrée : 784 neurones (pixels 28×28 aplatis)
- Couche cachée : 128 neurones avec activation ReLU
- Couche de sortie : 10 neurones (un par classe de chiffre)


````python
class SimpleNN(nn.Module):
    def __init__(self):
        super(SimpleNN, self).__init__()

        # Définir les couches
        self.flatten = nn.Flatten()           # Aplatir 28x28 en 784
        self.fc1 = nn.Linear(784, 128)        # Première couche entièrement connectée
        self.fc2 = nn.Linear(128, 10)         # Couche de sortie

    def forward(self, x):
        """Passe avant à travers le réseau"""
        x = self.flatten(x)          # Aplatir : (batch, 1, 28, 28) → (batch, 784)
        x = self.fc1(x)              # Première couche
        x = F.relu(x)                # Activation ReLU
        x = self.fc2(x)              # Couche de sortie
        return x                     # Logits bruts (softmax appliqué dans la perte)

# Créer le modèle et le déplacer vers le périphérique
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = SimpleNN().to(device)
print(model)
print(f"\nLe modèle est sur : {next(model.parameters()).device}")
````

**Comprendre l'architecture :**

1. **Flatten** : convertit une image 2D en vecteur 1D
2. **fc1** : apprend 128 caractéristiques à partir de l'entrée
3. **ReLU** : ajoute de la non-linéarité
4. **fc2** : associe les caractéristiques à 10 scores de classe


### 9.4 Définir la fonction de perte et l'optimiseur

**Fonction de perte :** CrossEntropyLoss (combine softmax et log-vraisemblance négative)
**Optimiseur :** SGD (descente de gradient stochastique)


````python
# Fonction de perte
criterion = nn.CrossEntropyLoss()

# Optimiseur : met à jour les poids du modèle selon les gradients
optimizer = optim.SGD(model.parameters(), lr=0.01, momentum=0.9)

print("Fonction de perte:", criterion)
print("Optimiseur :", optimizer)
````

### 9.5 Boucle d'entraînement

Le processus d'entraînement :
1. **Passe avant** : calculer les prédictions
2. **Calculer la perte** : dans quelle mesure le modèle se trompe-t-il ?
3. **Passe arrière** : calculer les gradients
4. **Mettre à jour les poids** : améliorer le modèle


````python
# Boucle d’entraînement
num_epochs = 3
print('Début de l’entraînement\n')

for epoch in range(num_epochs):
    model.train()  # Mettre le modèle en mode entraînement
    running_loss = 0.0

    for batch_idx, (images, labels) in enumerate(train_loader):
        # Déplacer les données vers le périphérique
        images, labels = images.to(device), labels.to(device)

        # Réinitialiser les gradients
        optimizer.zero_grad()

        # Passe avant
        outputs = model(images)
        loss = criterion(outputs, labels)

        # Passe arrière et optimisation
        loss.backward()
        optimizer.step()

        running_loss += loss.item()

        # Afficher la progression
        if batch_idx % 200 == 0:
            print(f'Époque [{epoch+1}/{num_epochs}], '
                  f'Étape [{batch_idx}/{len(train_loader)}], '
                  f'Perte : {loss.item():.4f}')

    # Afficher les statistiques de l’époque
    avg_loss = running_loss / len(train_loader)
    print(f'Époque [{epoch+1}/{num_epochs}] Perte moyenne : {avg_loss:.4f}\n')

print('Finished Training!')
````

### 9.6 Évaluer le modèle

Testons notre modèle entraîné sur l'ensemble de test !


````python
# Évaluation
model.eval()  # Mettre le modèle en mode évaluation
correct = 0
total = 0

with torch.no_grad():  # Désactiver le calcul des gradients
    for images, labels in test_loader:
        images, labels = images.to(device), labels.to(device)

        outputs = model(images)
        _, predicted = torch.max(outputs.data, 1)  # Obtenir la classe au score le plus élevé

        total += labels.size(0)
        correct += (predicted == labels).sum().item()

accuracy = 100 * correct / total
print(f'Précision sur 10 000 images de test : {accuracy:.2f}%')
print(f'Prédictions correctes : {correct}/{total}')
````

## 10. Résumé et bonnes pratiques

### Points essentiels

1. Les **tenseurs** sont la structure de données fondamentale de PyTorch
   - Similaires aux tableaux NumPy, mais compatibles GPU
   - Prennent en charge la différentiation automatique

2. **Autograd** calcule automatiquement les gradients
   - Utilisez `requires_grad=True` pour suivre les opérations
   - Appelez `.backward()` pour calculer les gradients
   - N'oubliez pas de réinitialiser les gradients !

3. Les **réseaux de neurones** sont construits avec `nn.Module`
   - Définissez les couches dans `__init__`
   - Implémentez la passe avant dans `forward()`
   - Utilisez les fonctions de perte et les optimiseurs intégrés

### Bonnes pratiques

#### 1. Gestion des périphériques
```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = model.to(device)
data = data.to(device)
```

#### 2. Structure de la boucle d'entraînement
```python
model.train()  # Training mode
for images, labels in train_loader:
    optimizer.zero_grad()      # Clear gradients
    outputs = model(images)    # Forward pass
    loss = criterion(outputs, labels)
    loss.backward()            # Backward pass
    optimizer.step()           # Update weights
```

#### 3. Mode évaluation
```python
model.eval()  # Evaluation mode
with torch.no_grad():  # Disable gradient computation
    outputs = model(test_data)
```

#### 4. Sauvegarder et charger des modèles
```python
# Save
torch.save(model.state_dict(), 'model.pth')

# Load
model.load_state_dict(torch.load('model.pth'))
```

### Erreurs courantes à éviter

❌ **Oublier de réinitialiser les gradients**
```python
# Wrong - gradients accumulate!
loss.backward()
optimizer.step()
```
✅ **Correct**
```python
optimizer.zero_grad()
loss.backward()
optimizer.step()
```

❌ **Modèle et données sur des périphériques différents**
```python
model = model.to('cuda')
output = model(data)  # data still on CPU - Error!
```
✅ **Correct**
```python
model = model.to(device)
data = data.to(device)
output = model(data)
```

❌ **Oublier d'appeler `.eval()` pendant les tests**
```python
# Wrong - dropout/batchnorm still active
with torch.no_grad():
    output = model(test_data)
```
✅ **Correct**
```python
model.eval()
with torch.no_grad():
    output = model(test_data)
```
